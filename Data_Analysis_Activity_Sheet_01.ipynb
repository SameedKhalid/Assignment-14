```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
ec = pd.read_csv('pakistani_ecommerce_dataset.csv')
st = pd.read_csv('student_lifestyle_dataset.csv')
print(ec.shape, st.shape)
```

## Dataset 1: Pakistani E-Commerce

### Q1: Most popular product category

```python
print(ec['product_category'].value_counts())
```

**Answer:** **Home Decor** (30 orders).

### Q2: Average order value

```python
print(round(ec['order_value'].mean(), 2))
```

**Answer:** **PKR 2,832.15**.

### Q3: City with the highest number of orders

```python
print(ec['city'].value_counts())
```

**Answer:** **Karachi** (24 orders).

### Q4: Most used payment method

```python
print(ec['payment_method'].value_counts())
```

**Answer:** **Cash on Delivery** (29 orders).

### Q5: Bar chart of average order value by city

```python
avg_city = ec.groupby('city')['order_value'].mean().sort_values(ascending=False)
print(avg_city.round(2))
plt.figure(figsize=(8,5))
sns.barplot(x=avg_city.index, y=avg_city.values, hue=avg_city.index, palette='viridis', legend=False)
plt.title('Average Order Value by City')
plt.xlabel('City'); plt.ylabel('Average Order Value (PKR)')
plt.show()
```

**Answer:** Islamabad has the highest average order value (about 3,012) and Rawalpindi the lowest (about 2,447).

## Dataset 2: Student Lifestyle & Study Habits

### Q1: Average study hours

```python
print(round(st['study_hours'].mean(), 2))
```

**Answer:** **4.4 hours** per day.

### Q2: Who sleeps more, males or females?

```python
print(st.groupby('gender')['sleep_hours'].mean().round(2))
```

**Answer:** **Females** sleep more (6.80 hrs vs 6.39 hrs for males).

### Q3: Relationship between screen time and sleep hours

```python
print('Correlation:', round(st['screen_time'].corr(st['sleep_hours']), 3))
sns.regplot(data=st, x='screen_time', y='sleep_hours')
plt.title('Screen Time vs Sleep Hours')
plt.show()
```

**Answer:** The correlation is about 0.05, so there is essentially **no relationship** between screen time and sleep hours.

### Q4: Histogram of daily transport cost

```python
plt.hist(st['daily_transport_cost'], bins=10, color='steelblue', edgecolor='black')
plt.title('Histogram of Daily Transport Cost')
plt.xlabel('Daily Transport Cost (PKR)'); plt.ylabel('Number of Students')
plt.show()
```

### Q5: Average internet usage by gender

```python
ig = st.groupby('gender')['internet_usage'].mean()
print(ig.round(2))
sns.barplot(x=ig.index, y=ig.values, hue=ig.index, palette='Set2', legend=False)
plt.title('Average Internet Usage by Gender'); plt.ylabel('Avg Internet Usage (hrs)')
plt.show()
```

**Answer:** **Males** use more internet (6.00 hrs vs 5.32 hrs for females).
